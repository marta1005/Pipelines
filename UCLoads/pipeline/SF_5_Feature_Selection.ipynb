#### 5.1 Feature analysis — input/output statistics

```python
%matplotlib inline
import matplotlib.pyplot as plt

ms = workflow.metadata.get_step_data(['metadata', 'Feature_Selection', 'Add_Remove_Features'])
inputs  = ms['inputs']
outputs = ms['outputs']

print("Inputs:",  inputs)
print("Outputs:", outputs)

# Correlation matrix — inputs vs outputs
import pandas as pd
import numpy as np

corr = Train_set[inputs + outputs].corr()
fig, ax = plt.subplots(figsize=(12, 10))
im = ax.imshow(corr.values, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(im, ax=ax)
ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.columns)))
ax.set_xticklabels(corr.columns, rotation=45, ha='right', fontsize=8)
ax.set_yticklabels(corr.columns, fontsize=8)
ax.set_title('Pearson Correlation — Inputs & Outputs')
plt.tight_layout()
plt.show()
plt.close()
```
