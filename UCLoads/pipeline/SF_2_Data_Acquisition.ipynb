Objectives:
- Load aerodynamic loads from `.mon` files (Tecplot format).

  - **MS_FUS_FTJ**: Net-Fy, Net-Fz, Net-Mx, Net-My, Net-Mz at fuselage frame

- Save raw merged dataset.

```python
from data_acquisition.load_mon import batch_extract
raw_data = batch_extract(workflow)
print(raw_data.shape)
raw_data.head()
```
