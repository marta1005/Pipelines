#### 8.1 Quick inference test — first row of training set

```python
import joblib, pandas as pd

ms = workflow.metadata.get_step_data(['metadata', 'Model_Selection'])
inputs  = ms['inputs']
outputs = ms['outputs']

Train_set = workflow.load_data(workflow.config['job_name'] + '_Train_set.csv')
sample = Train_set[inputs].iloc[[0]]

pipelines = workflow.metadata.get_step_data(['metadata', 'Model_Deployment', 'pipelines'])
for p in pipelines:
    pipeline = joblib.load(p['pipeline_file'])
    pred = pipeline.predict(sample)
    pred_df = pd.DataFrame(pred, columns=outputs)
    print(f"\n[{p['label']}] prediction on first row:")
    display(pred_df)
    print(f"  True:")
    display(Train_set[outputs].iloc[[0]])
```
