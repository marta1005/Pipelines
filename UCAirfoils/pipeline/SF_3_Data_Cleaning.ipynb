# Surrogate Factory — UCAirfoils
## Chapter 3. Data Cleansing
Objectives:
- Drop points NeuralFoil could not analyse reliably (`analysis_confidence`).
- Fill any remaining missing values.

The confidence filter matters: low-confidence points carry roughly **2.5x the
CD spread** of the rest, so leaving them in trains the surrogate on noise.


### 0. Workflow initialisation

```python
from IPython.display import display, HTML, JSON
from surrogate_factory.workflow import Workflow

workflow = Workflow("pipeline_config.yaml")
workflow.resume()
```

### 3. Data Cleansing

```python
workflow.import_metadata(stage_name="SF_3_Data_Cleansing")
```

```python
Dataset = workflow.load_data(workflow.config["job_name"] + "_Raw.csv")
Dataset.describe()
```

#### 3.1 Filter by NeuralFoil confidence

```python
from data_cleansing.confidence import filter_by_confidence
Dataset = filter_by_confidence(workflow, Dataset)
```

#### 3.2 Missing values

```python
from data_cleansing.manage_missing import replace_missing_values
Dataset_cleaned = replace_missing_values(workflow, Dataset)
print("Missing values after fill:")
print(Dataset_cleaned.isna().sum())
```

### Save
