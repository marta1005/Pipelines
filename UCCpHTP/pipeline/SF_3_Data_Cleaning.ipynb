### 3. Data Cleansing — null check only

```python
workflow.import_metadata(stage_name="SF_3_Data_Cleansing")
```

```python
job = workflow.config['job_name']
Train_set = workflow.load_data(job + '_Train_set.csv')

nulls = Train_set.isnull().sum()
remaining = nulls[nulls > 0]
if remaining.empty:
    print('No missing values — data is clean.')
else:
    print('WARNING: missing values found:')
    print(remaining)

Train_set.describe()
```
